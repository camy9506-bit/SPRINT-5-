### 2.3 Corregir formatos numéricos y de fecha

**🎯Objetivo:**
Asegurar que las columnas de fechas y valores numéricos estén en formatos correctos para permitir análisis, cálculos y comparaciones precisas.

**Instrucciones:**

- Convierte las columnas de fecha de `traffic` a formato `datetime`. Haz el cambio a prueba de errores.
- En el dataset `eco`, limpia los valores numéricos:
    - En `city_gdp_capita`: elimina separadores de miles (`.`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `unemployment_pct`: elimina el símbolo de porcentaje (`%`) y reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
    - En `population_m`: reemplaza las comas (`','`) por puntos (`'.'`) antes de convertir a tipo `float`.
- Finalmente, crea una nueva columna llamada `population` multiplicando `population_m` por 1,000,000 para obtener la población total.


<details>
<summary>Haz clic para ver la pista</summary>
para eliminar símbolos, puedes reemplazarlos por un texto vacío.

```python
# Convertir las columnas de traffic a tipo fecha con pd.to_datetime()
import pandas as pd
traffic['update_time_utc'] = pd.to_datetime(traffic["update_time_utc"], errors="coerce", utc=True)
traffic['update_time_utc_week_ago'] = pd.to_datetime(traffic["update_time_utc_week_ago"], errors="coerce", utc=True)

# Verificar el cambio de tipo de dato
traffic.info()
```

```python


import pandas as pd

eco = pd.read_csv('/datasets/oecd_city_economy.csv')

# 1. Estandarizar nombres: pasamos a minúsculas, cambiamos espacios/barras por guiones, y BORRAMOS los paréntesis
eco.columns = (
    eco.columns.str.lower()
    .str.replace(' ', '_', regex=False)
    .str.replace('/', '_', regex=False)
    .str.replace('(', '', regex=False)
    .str.replace(')', '', regex=False)
)
# ¡Ahora tus columnas se llaman exactamente: 'city_gdp_capita', 'unemployment_%' y 'population_m'!

# 2. Limpiar separadores de GDP por cápita
eco['city_gdp_capita'] = (
    eco['city_gdp_capita']
    .astype(str)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
eco['city_gdp_capita'] = pd.to_numeric(eco['city_gdp_capita'], errors='coerce')

# 3. Limpiar desempleo (usando 'unemployment_%' que es el nombre real tras el renombrado)

eco['unemployment_pct'] = (
    eco['unemployment_%']
    .astype(str)
    .str.replace('%', '', regex=False)    # Quitamos el porcentaje primero
    .str.replace(',', '.', regex=False)   # Cambiamos comas por puntos decimales
)
eco['unemployment_pct'] = pd.to_numeric(eco['unemployment_pct'], errors='coerce')

# 4. Limpiar separadores de población en millones 
eco['population_m'] = (
    eco['population_m']
    .astype(str)
    .str.replace('.', '', regex=False)
    .str.replace(',', '.', regex=False)
)
eco['population_m'] = pd.to_numeric(eco['population_m'], errors='coerce')

# Cambiamos la coma por el punto decimal
eco['pm2.5_μg_m³'] = (
    eco['pm2.5_μg_m³']
    .astype(str)
    .str.replace(',', '.', regex=False) 
)

# Convertir la columna a tipo numérico
eco['pm2.5_μg_m³'] = pd.to_numeric(eco['pm2.5_μg_m³'], errors='coerce')

# Eliminar la columna de porcentaje
eco = eco.drop(columns=['unemployment_%'])

# Verificar que el cambio se aplicó correctamente
eco[['city', 'pm2.5_μg_m³']].head(3)

eco[['unemployment_pct', 'pm2.5_μg_m³']].isna().sum()


# 5. Calcular la población total en unidades absolutas columna llamada population
eco['population'] = eco['population_m'] * 1000000


# 6. Verificar que todo quedó perfecto
eco.info()
eco.head()

```

## 🧩Paso 4: Analizar y resumir datos de movilidad

Como el dataset de tráfico contiene **múltiples registros por ciudad**. En esta parte, calcularás los promedios anuales por ciudad para simplificar el análisis y obtener una visión más clara de las tendencias generales.

### 4.1 Calcular promedios de tráfico por ciudad

**🎯Objetivo:**
Obtener una vista consolidada del tráfico promedio por ciudad y año, para analizar patrones generales sin depender de datos diarios.

**Instrucciones**

- Agrupa los datos por `city`, `country` y `year`.
- Calcula el promedio **solo de las métricas de tráfico más relevantes**: como `jams_delay`, `traffic_index_live`, `jams_length_kms`, `jams_count`, `mins_delay`, y tiempos de viaje (`travel_time_live_per_10kms_mins` y `travel_time_hist_per_10kms_mins`).
- Guarda el resultado como `traffic_city_year_2024`, mantén las columnas como variables (no índices).


<details>
<summary>Haz clic para ver la pista</summary>
Usa ".agg()" para aplicar funciones de promedio. Al final, reinicia el índice para mantener las columnas de la agrupación como variables (no índices).

```python
# Calcular los  promedios de trafico por ciudad, país y año
traffic_city_year_2024 = traffic_2024.groupby(['city', 'country', 'year']).agg({
    'jams_delay': 'mean',
    'traffic_index_live': 'mean',
    'jams_length_in_km': 'mean',
    'jams_count': 'mean',
    'mins_delay': 'mean',
    'travel_time_live_per_10kms_mins': 'mean',  
    'travel_time_historic_per_10kms_mins': 'mean'
}).reset_index()

# Mostrar resultado
traffic_city_year_2024.head()
```

```python
import pandas as pd

# =====================================================================
# PASO A: REGENERAR LA AGRUPACIÓN DE TRÁFICO (Limpia la memoria de Pandas)
# =====================================================================

traffic_city_year_2024 = traffic_2024.groupby(['city', 'country', 'year']).agg({
    'jams_delay': 'mean',
    'traffic_index_live': 'mean',       
    'jams_length_in_km': 'mean',
    'jams_count': 'mean',
    'mins_delay': 'mean',
    'travel_time_live_per_10kms_mins': 'mean',
    'travel_time_historic_per_10kms_mins': 'mean'
}).reset_index()

# =====================================================================
# PASO B: ESTANDARIZACIÓN EXACTA PARA SNAKE_CASE (Tráfico)
# =====================================================================
traffic_city_year_2024.columns = (
    traffic_city_year_2024.columns.str.lower()
    .str.replace('jams', 'jams_', regex=False)
    .str.replace('traffic', 'traffic_', regex=False)
    .str.replace('mins', 'mins_', regex=False)
    .str.replace('index', 'index_', regex=False)
    .str.replace('lengthinkms', 'length_kms', regex=False)
    .str.replace('__', '_', regex=False)
)

# Forzamos el nombre exacto de las columnas de tiempo de viaje que pide left_cols
traffic_city_year_2024 = traffic_city_year_2024.rename(columns={
    'traveltimeliveper10kmsmins_': 'travel_time_live_per_10kms_mins',
    'traveltimehistoricper10kmsmins_': 'travel_time_hist_per_10kms_mins'
})

# =====================================================================
# PASO C: CORRECCIÓN DE CONTAMINACIÓN (Economía)
# =====================================================================
eco_2024.columns = eco_2024.columns.str.replace('pm2.5_μg_m³', 'pm25', regex=False)

# =====================================================================
# PASO D: ORDEN SOLICITADO Y MERGE FINAL
# =====================================================================
left_cols = [
    'city', 'country', 'year', 'jams_delay', 'traffic_index_live',
    'jams_length_in_km', 'jams_count', 'mins_delay'
]

right_cols = ['city', 'year', 'city_gdp_capita', 'unemployment_pct', 'pm25', 'population']

# Crear datasets reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# Unir datasets usando un inner join
merged = pd.merge(traffic_2024_small, eco_2024_small, on=['city', 'year'], how='inner')

# Crear la variable top_10 para tus gráficos
top_10 = merged.sort_values(by='city_gdp_capita', ascending=False).head(10)

# Mostrar el resultado final de la unión
merged.head(5)
```
